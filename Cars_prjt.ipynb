```python
#!pip install wikipedia
import wikipedia

def get_car_specs(car_name):
    try:
        # Search for car on Wikipedia
        wikipedia.set_lang("en")
        search_results = wikipedia.search(car_name)

        if not search_results:
            print("\nNo car found with that name. Try again.")
            return

        # Get the first matching page
        page = wikipedia.page(search_results[0])

        print("\nCar Specifications (from Wikipedia):")
        print(f"Title: {page.title}\n")
        print(page.summary.split('\n')[0])  # Print only first paragraph

        print("\nMore info:")
        print(page.url)

    except wikipedia.exceptions.DisambiguationError as e:
        print("\nMultiple results found. Try to be more specific. Examples:")
        for option in e.options[:5]:
            print("-", option)
    except wikipedia.exceptions.PageError:
        print("\nPage not found. Try again with a different car name.")
    except Exception as e:
        print("\nAn error occurred:", e)

# Take input
car_name = input("Enter the car name: ")
get_car_specs(car_name)
```
