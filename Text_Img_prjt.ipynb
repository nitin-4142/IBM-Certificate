```python
# Install libraries
!pip install -q diffusers transformers accelerate gradio

import torch
from diffusers import StableDiffusionPipeline
import gradio as gr

# Load a fast model (Stable Diffusion Turbo)
pipe = StableDiffusionPipeline.from_pretrained(
    "stabilityai/sd-turbo",
    torch_dtype=torch.float16,
    variant="fp16",
).to("cuda")

# Function to convert text to image
def text_to_image(prompt):
    with torch.no_grad():
        with torch.autocast("cuda"):
            image = pipe(prompt=prompt, num_inference_steps=2, guidance_scale=0.0).images[0]
    return image

# Create Gradio app
gr.Interface(
    fn=text_to_image,
    inputs=gr.Textbox(placeholder="Enter a description..."),
    outputs="image",
    title="Text to Image Generator",
    description="Enter a prompt and generate an image!",
).launch()
```
