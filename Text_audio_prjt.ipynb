```python
# Step 1: Install gTTS and Gradio
!pip install -q gTTS gradio

# Step 2: Import libraries
from gtts import gTTS
import gradio as gr

# Step 3: Define text-to-audio function
def text_to_audio(text):
    tts = gTTS(text=text, lang='en')
    output_path = "output.mp3"
    tts.save(output_path)
    return output_path

# Step 4: Build Gradio app
gr.Interface(
    fn=text_to_audio,
    inputs=gr.Textbox(placeholder="Type your text here..."),
    outputs=gr.Audio(type="filepath"),
    title="Fast Text to Speech (gTTS)",
    description="Enter text and instantly get the spoken audio!",
    theme="default",
).launch()
```
