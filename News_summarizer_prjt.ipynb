```python
# Import required libraries
from transformers import pipeline
import gradio as gr

# Load pre-trained summarization model (BART or T5)
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Define function to summarize news articles
def summarize_news(news_text):
    summary = summarizer(news_text, max_length=200, min_length=50, do_sample=False)
    return summary[0]['summary_text']

# Create Gradio interface for summarization
gr.Interface(
    fn=summarize_news,
    inputs=gr.Textbox(lines=15, placeholder="Enter news article here..."),
    outputs="text",
    title="📰 AI News Summarizer",
    description="Enter a news article or long text, and the AI will provide a summary.",
    theme="default",
).launch()
```
