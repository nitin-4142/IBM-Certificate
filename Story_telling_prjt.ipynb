```python
# Import necessary libraries
from transformers import pipeline
import gradio as gr

# Load GPT-2 text generation pipeline from Hugging Face
story_generator = pipeline("text-generation", model="gpt2")

# Function to generate a story based on the prompt
def tell_story(prompt):
    generated_story = story_generator(prompt, max_length=300, num_return_sequences=1)
    return generated_story[0]['generated_text']

# Create a simple Gradio interface for user input and output
gr.Interface(
    fn=tell_story,
    inputs=gr.Textbox(lines=2, placeholder="Enter your story prompt here..."),
    outputs="text",
    title="🤖 Storytelling Chatbot",
    description="Type a prompt, and let the AI create a story for you!",
    theme="default",
).launch()
```
