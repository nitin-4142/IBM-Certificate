```python
# Install required packages
!pip install youtube-transcript-api transformers gradio

from youtube_transcript_api import YouTubeTranscriptApi
from transformers import pipeline
import gradio as gr

# Function to fetch transcript from YouTube
def get_transcript(video_url):
    try:
        video_id = video_url.split('v=')[-1].split('&')[0]
        transcript_list = YouTubeTranscriptApi.get_transcript(video_id)
        transcript = " ".join([x['text'] for x in transcript_list])
        return transcript
    except Exception as e:
        return f"Error fetching transcript: {str(e)}"

# Load summarization model
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Final function to get summary
def summarize_youtube(video_url):
    transcript = get_transcript(video_url)
    if transcript.startswith("Error"):
        return transcript

    # Due to model input limits (~1024 tokens), we truncate long transcripts
    max_length = 1000
    transcript = transcript[:max_length]

    summary = summarizer(transcript, max_length=130, min_length=30, do_sample=False)[0]['summary_text']
    return summary

# Gradio UI
gr.Interface(
    fn=summarize_youtube,
    inputs=gr.Textbox(placeholder="Paste YouTube video URL here..."),
    outputs="text",
    title="YouTube Video Summarizer",
    description="Paste a YouTube video link and get a quick summary of its transcript!"
).launch()
```
